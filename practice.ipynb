```python
def armstrong(n):
    temp = n
    digits = len(str(n))
    arm = 0

    while temp > 0:
        digit = temp % 10
        arm += digit ** digits
        temp = temp // 10

    if arm == n:
        print("Armstrong")
    else:
        print("Not Armstrong")

armstrong(153)
```

```python
# Given an array nums of size n which may contain duplicate elements.
# Rreturn a list of pairs where each pair contains a unique element from the array and its frequency in the array.
# You may return the result in any order, but each element must appear exactly once in the output.
# Example 1
# Input: nums = [1, 2, 2, 1, 3]
# Output: [[1, 2], [2, 2], [3, 1]]

def arrcount(arr):
    freq = {}

    for x in arr:
        if x in freq:
            freq[x] += 1
        else:
            freq[x] = 1

    result = []

    for key, value in freq.items():
        result.append([key, value])

    return result

print(arrcount([1, 2, 2, 1, 3]))
```

```python
#Given a binary array nums, return the maximum number of consecutive 1s in the array.
def maxconsecutive(arr):
  max_count = 0
  current_count = 0
  for num in arr:
    if num == 1:
      current_count += 1
    else:
      # If a 0 is encountered, reset current_count
      max_count = max(max_count, current_count)
      current_count = 0
  # After the loop, update max_count one last time in case the array ends with 1s
  max_count = max(max_count, current_count)
  return max_count

print(maxconsecutive([1, 1, 0, 0, 1, 1, 1, 0]))
print(maxconsecutive([1, 1, 1, 1, 0, 1]))
print(maxconsecutive([0, 0, 0]))
print(maxconsecutive([1, 1, 1]))
```

```python
#Given an array nums of size n and an integer k, find the length of the longest sub-array that sums to k. If no such sub-array exists, return 0.

def longestarray(arr, k):
    max_len = 0

    for i in range(len(arr)):
        total = 0

        for j in range(i, len(arr)):
            total += arr[j]

            if total == k:
                max_len = max(max_len, j - i + 1)

    return max_len

print(longestarray([10, 5, 2, 7, 1, 9], 15))

```

```python
#Given an array of integers nums and an integer target. Return the indices(0 - indexed) of two elements in nums such that they add up to target.
def twoSum(nums, target=8):
    for i in range(len(nums)):
        result = []
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                result.append(i)
                result.append(j)
                return result

print(twoSum([1, 6, 2, 10, 3]))

```

```python
#Given an array arr of n integers, where arr[i] represents price of the stock on the ith day. Determine the maximum profit achievable by buying and selling the stock at most once.
class Solution:
    def stockBuySell(self, arr, n):
        min_price = arr[0]
        max_profit = 0

        for i in range(1, n):
            profit = arr[i] - min_price
            max_profit = max(max_profit, profit)

            min_price = min(min_price, arr[i])

        return max_profit


s = Solution()
print(s.stockBuySell([10, 7, 5, 8, 11, 9], 6))
```

```python
'''Given an integer array nums of even length consisting of an equal number of positive and negative integers.Return the answer array in such a way that the given conditions are met:
Every consecutive pair of integers have opposite signs.
For all integers with the same sign, the order in which they were present in nums is preserved.
The rearranged array begins with a positive integer.'''
def rearrange(nums):
    positive = []
    negative = []
    result = []

    for num in nums:
        if num > 0:
            positive.append(num)
        else:
            negative.append(num)

    for i in range(len(positive)):
        result.append(positive[i])
        result.append(negative[i])

    return result


print(rearrange([2, 4, 5, -1, -3, -4]))
```

```python
#A leader in an array is an element whose value is strictly greater than all elements to its right in the given array. The rightmost element is always a leader. The elements in the leader array must appear in the order they appear in the nums array.
def ladder(arr):
    result = []

    for i in range(len(arr)):
        leader = True

        for j in range(i + 1, len(arr)):
            if arr[i] <= arr[j]:
                leader = False
                break

        if leader:
            result.append(arr[i])

    return result
print(ladder([1, 2, 5, 3, 1, 2]))



```

```python
#Return the length of the longest sequence of consecutive integers. The integers in this sequence can appear in any order.

def longestsequence(arr):
    arr.sort()

    count = 1
    max_count = 1

    for i in range(1, len(arr)):
        if arr[i] == arr[i-1] + 1:
            count += 1
        elif arr[i] != arr[i-1]:
            count = 1

        max_count = max(max_count, count)

    return max_count


print(longestsequence([100, 4, 200, 1, 3, 2]))
```

```python
#Given an m x n integer matrix matrix, if an element is 0, set its entire row and column to 0. You must do it in place.
def zeromatrix(arr):
    rows = []
    cols = []

    for i in range(len(arr)):
        for j in range(len(arr[0])):
            if arr[i][j] == 0:
                rows.append(i)
                cols.append(j)

    for r in rows:
        for j in range(len(arr[0])):
            arr[r][j] = 0

    for c in cols:
        for i in range(len(arr)):
            arr[i][c] = 0

    return arr


print(zeromatrix([[1,1,1],[1,0,1],[1,1,1]]))
```

```python
#Print the matrix in spiral manner
def spiral(matrix):
    top = 0
    bottom = len(matrix) - 1
    left = 0
    right = len(matrix) - 1

    while top <= bottom and left <= right:

        # Left to Right
        for i in range(left, right + 1):
            print(matrix[top][i], end=" ")
        top += 1

        # Top to Bottom
        for i in range(top, bottom + 1):
            print(matrix[i][right], end=" ")
        right -= 1

        # Right to Left
        if top <= bottom:
            for i in range(right, left - 1, -1):
                print(matrix[bottom][i], end=" ")
            bottom -= 1

        # Bottom to Top
        if left <= right:
            for i in range(bottom, top - 1, -1):
                print(matrix[i][left], end=" ")
            left += 1


spiral([[1, 2, 3],
        [4, 5, 6],
        [7, 8, 9]])
```

```python
#VGiven an array of integers nums and an integer k, return the total number of subarrays whose sum equals to k.
def subarraySum(nums, k):
    count = 0

    for i in range(len(nums)):
        total = 0

        for j in range(i, len(nums)):
            total += nums[j]

            if total == k:
                count += 1

    return count


print(subarraySum([1, 1, 1], 2))
```

```python
#Given a string s, representing a large integer, the task is to return the largest-valued odd integer (as a string) that is a
#substring of the given string s.The number returned should not have leading zero's. But the given input string may have leading zero.
 #(If no odd number is found, then return empty string.)
 #Input : s = "5347" Output : "5347"
def largestval(s):
    for i in range(len(s) - 1, -1, -1):
        if int(s[i]) % 2 != 0:
            return s[:i + 1]
    return ""

print(largestval("5347"))   # 5347
print(largestval("4206"))   # ""
print(largestval("05234"))  # 0523
```
